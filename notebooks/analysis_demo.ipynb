# Analysis Demo

Demonstrates aggregation and visualisation utilities using synthetic multilingual corpus data.

```python
import pandas as pd
import numpy as np
from multilingual_topic.analysis import (
    volume_over_time,
    top_n_distribution,
    platform_language_breakdown,
    Heatmap,
    top_accounts_by_country,
)
```

## Generate dummy data

```python
np.random.seed(42)
n = 2000

countries  = ['Egypt', 'Indonesia', 'Turkey', 'Pakistan', 'Saudi Arabia', 'Morocco', 'Iraq', 'India', 'Iran']
themes     = ['Geopolitics', 'Religious values', 'Child welfare', 'Media freedom', 'National identity', 'Diplomacy']
platforms  = ['Twitter', 'Facebook', 'Instagram', 'Telegram', 'YouTube']
languages  = ['Arabic', 'English', 'Urdu', 'Turkish', 'Indonesian', 'Persian', 'Hindi']
accounts   = [f'account_{i}' for i in range(80)]

df = pd.DataFrame({
    'date':     pd.date_range('2021-01-01', periods=n, freq='6h') + pd.to_timedelta(np.random.randint(0, 6, n), unit='h'),
    'country':  np.random.choice(countries, n, p=[0.18, 0.15, 0.14, 0.13, 0.12, 0.10, 0.08, 0.06, 0.04]),
    'theme':    np.random.choice(themes, n),
    'platform': np.random.choice(platforms, n, p=[0.35, 0.25, 0.15, 0.15, 0.10]),
    'language': np.random.choice(languages, n, p=[0.30, 0.20, 0.15, 0.12, 0.10, 0.08, 0.05]),
    'account':  np.random.choice(accounts, n),
    'engagement': np.random.exponential(scale=50, size=n).astype(int),
})

print(f'{len(df)} posts | {df.country.nunique()} countries | {df.theme.nunique()} themes')
df.head()
```

## 1. Volume over time by theme

```python
fig = volume_over_time(df, date_col='date', group_col='theme', top_n=6, title='Monthly post volume by theme')
fig.show()
```

## 2. Top country distribution

```python
fig = top_n_distribution(df, col='country', top_n=9, title='Post volume by country')
fig.show()
```

## 3. Platform × language breakdown

```python
breakdown = platform_language_breakdown(df, platform_col='platform', language_col='language')
breakdown
```

## 4. Country × theme heatmap

```python
hm = Heatmap(df, row_col='country', col_col='theme')
fig = hm.plot(title='Country × Theme distribution')
fig.show()
```

## 5. Top accounts by country

```python
account_pivot = top_accounts_by_country(df, account_col='account', country_col='country', top_n=10)
account_pivot
```
